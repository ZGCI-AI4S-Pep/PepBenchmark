```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import json 
import pandas as pd
from constants import data_dir, meta_data_dir, blues, reds, output_dir, font_prop
```

```python
os.makedirs(save_dir := (output_dir / Path("seq_len")), exist_ok=True)
```

```python
with open(meta_data_dir / Path("task_types.json"), "r") as f:
    task_types = json.load(f)
dataset_names = list(task_types.keys())
```

```python
for dataset_name in dataset_names:
    data_info = pd.read_csv(data_dir / Path(dataset_name) / Path("enhanced_data.csv"))
    lengths = data_info["length"].values

    len_values, counts = np.unique(lengths, return_counts=True)
    mean = np.mean(lengths)
    med = np.median(len_values)

    if len(counts) > 10:
        plt.figure(figsize=(16, 6))
        plt.bar(
            len_values, 
            counts, 
            color=blues[3], 
            width=1.0, 
            align='edge',
            edgecolor=blues[-2],
            alpha=0.8
        )
    else:
        plt.figure(figsize=(6, 6))
        plt.bar(
            len_values, 
            counts, 
            color=blues[3], 
            width=1.0, 
            align='edge',
            edgecolor=blues[-2],
            alpha=0.8
        )
        int_med = int(med)
        plt.xlim(max(0, int_med-10), min(int_med+10, 20))
        ticks = plt.xticks()[0]
        new_labels = [int(t) for t in ticks]
        plt.xticks(ticks, new_labels)

    plt.axvline(x=mean, color=reds[-3], linestyle='--', linewidth=2)
    plt.axvline(x=med, color=reds[-3], linestyle=':', linewidth=2)

    plt.xlabel("Number of Amino Acids (Sequence Length)", fontsize=14, fontproperties=font_prop,)
    plt.ylabel("Number of Peptides", fontsize=14, fontproperties=font_prop,)
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)

    plt.savefig(
        save_dir / Path(f"{dataset_name}.png"), 
        dpi=200, 
        bbox_inches="tight", 
        transparent=True
    )
```
