```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import json
import pandas as pd
from collections import Counter
from scipy.stats import gaussian_kde
from copy import deepcopy
from constants import data_dir, meta_data_dir, blues, reds, output_dir, font_prop
```

```python
data_dir = "/Users/zhoukuangqi/Desktop/pepbenchmark/0923_ncgen"
```

```python
colors = {
    "train": reds[3],
    "pred": blues[2],
}
edge_colors = {
    "train": reds[-2],
    "pred": blues[-2],
}
legends = {
    "train": "training data",
    "pred": "predictions",
}

x_axis_names = {
    "molwt": "Mocule Weight",
    "logp": "Log P",
    "nhd": "Number of HBond Donors",
    "nha": "Number of HBond Acceptors",
    "tpsa": "TPSA",
}
col_names = {
    "molwt": "MolWt",
    "logp": "MolLogP",
    "nhd": "NumHDonors",
    "nha": "NumHAcceptors",
    "tpsa": "TPSA",
}
prop_names = list(x_axis_names.keys())
```

```python
os.makedirs(save_dir := (output_dir / Path("ncgen_0923")), exist_ok=True)
```

```python
props = {prop_name: {"train": [], "pred": []} for prop_name in prop_names}

data_info = pd.read_csv(
    data_dir / Path("training_physchem_properties.csv")
)
for prop_name in prop_names:
    props[prop_name]["train"] = data_info[col_names[prop_name]].values

data_info = pd.read_csv(
    data_dir / Path("predicted_physchem_properties.csv")
)
for prop_name in prop_names:
    props[prop_name]["pred"] = data_info[col_names[prop_name]].values
```

## Figure 1

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
for i, prop_name in enumerate(["molwt", "logp", "tpsa"]):
    for split in ['train', 'pred']:
        prop_values = np.asarray(props[prop_name][split])
        p_prop = gaussian_kde(prop_values)

        counts, bins, _ = axes[i].hist(
            prop_values, 
            bins=50, 
            alpha=0.5, 
            color=colors[split],
            density=True,
            label=legends[split],
        )
        legend_font_prop = deepcopy(font_prop)
        legend_font_prop.set_size(14)
        axes[i].legend(fontsize=20, prop=legend_font_prop)
        
        bin_width = bins[1] - bins[0]
        y_min, y_max = prop_values.min(), prop_values.max()
        x = np.linspace(y_min, y_max, 500)
        # y = p_prop(x) * len(prop_values) * bin_width
        y = p_prop(x)

        axes[i].plot(
            x,
            y,
            color=edge_colors[split],
            linewidth=2,
            alpha=0.8,
        )

    axes[i].set_xlabel(x_axis_names[prop_name], fontsize=14, fontproperties=font_prop)
    if i == 0:
        axes[i].set_ylabel("Frequency", fontsize=14, fontproperties=font_prop)
    axes[i].tick_params(axis='x', labelsize=13)
    axes[i].tick_params(axis='y', labelsize=13)
    axes[i].grid(True, linestyle='--', color='gray', alpha=0.5)
    plt.savefig(
        save_dir / Path(f"fig1.png"),
        dpi=200,
        bbox_inches="tight",
        transparent=True,
    )
```
