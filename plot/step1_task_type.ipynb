```python
import os
from pathlib import Path
import pandas as pd
import json
from constants import data_dir, meta_data_dir
```

```python
datasets = os.listdir(data_dir)
print(f"{(num_datasets:=len(datasets))=}")
```

```python
task_types = {}
```

```python
for dataset_name in datasets:
    info_fpath = data_dir / Path(dataset_name) / Path("enhanced_data.csv")
    data_info = pd.read_csv(info_fpath)

    
    labels = set(data_info["label"].unique().astype(int).tolist())

    if dataset_name.endswith("ic50"):
        break
    
    task_types[dataset_name] = "classification" if labels == {0, 1} else "regression"
```

```python
with open(meta_data_dir / "task_types.json", "w") as f:
    json.dump(task_types, f, indent=4, sort_keys=True)
```
