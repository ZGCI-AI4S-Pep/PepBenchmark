# Cluster Module Tutorial

This notebook demonstrates the `pepbenchmark.cluster` module for sequence clustering.

```python
import numpy as np
from pepbenchmark.cluster import create_clusterer, list_available_methods

print("Setup complete!")
print(f"Available methods: {list_available_methods()}")
```

```python
# Create sample sequences
sequences = [
    "ACDEFGHIKLMNPQRSTVWY",
    "ACDEFGHIKLMNPQRSTVWY",
    "ACDEFGHIKLMNPQRSTV",
    "MVHLTSQPAK",
    "MVHLTSQPAK",
]

# Create similarity matrix
sim_matrix = np.array([
    [1.00, 0.95, 0.80, 0.15, 0.12],
    [0.95, 1.00, 0.78, 0.14, 0.95],
    [0.80, 0.78, 1.00, 0.10, 0.12],
    [0.15, 0.14, 0.10, 1.00, 0.92],
    [0.12, 0.95, 0.12, 0.92, 1.00],
])

print(f"Sequences: {len(sequences)}")
print(f"Similarity matrix shape: {sim_matrix.shape}")
```

```python
# Perform clustering
clusterer = create_clusterer(method="similarity", similarity_threshold=0.80)
result = clusterer.cluster_sequences(sequences, similarity_matrix=sim_matrix)

print(f"Total clusters: {result.total_clusters}")
print(f"Assignments: {result.cluster_assignments}")

# Get statistics
stats = result.get_statistics()
print(f"\nStatistics:")
for key, value in stats.items():
    print(f"  {key}: {value}")
```
