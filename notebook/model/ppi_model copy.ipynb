```python
from scipy.datasets import face

from pepbenchmark.model.ppi_model import build_ppi_model
from pepbenchmark.dataset_manager.ppi_dataset import PPIDatasetManager

dataset = PPIDatasetManager(
    "PpI_ba",
    official_feature_names=["prot_fasta", "pep_fasta", "pep_ecfp6","label"],
    dataset_dir="../../PepBenchData/PepBenchData-50",
)
runner = build_ppi_model(
    model_type="ppi",
    dataset=dataset,
    protein_encoding="plm",
    peptide_encoding="plm",
    protein_model="facebook/esm2_t30_150M_UR50D",
    peptide_model="facebook/esm2_t30_150M_UR50D",
    freeze_protein_encoder=True,
    freeze_peptide_encoder=False,
    task_type="binary_classification",
    epochs=50,
    batch_size=64,
    learning_rate=5e-5,
    early_stopping_patience=5,
)

multi_results = runner.run_multi(
    run_seeds=[42, 43, 44, 45, 46],
    fold_seeds=[0, 1, 2, 3, 4],
    split_type="protein_mmseqs_cold_split",   
)
print(multi_results.to_dataframe())
summary = multi_results.get_summary_stats()
print(summary["test_roc-auc"])
```
