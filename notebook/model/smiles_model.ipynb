# 1. non-canonical peptide classification

```python


from pepbenchmark.model.smiles_model import build_smiles_model
from pepbenchmark.dataset_manager.single_dataset import SinglePeptideDatasetManager



dataset = SinglePeptideDatasetManager(
    "nc-antifungal",
    official_feature_names=["smiles", "label"],
    dataset_dir="../../PepBenchData/PepBenchData-50",
)

# 1) 标准模式
runner_std = build_smiles_model(
    dataset=dataset,
    model_name="seyonec/ChemBERTa-zinc-base-v1",
    task_type="binary_classification",
    smiles_key="smiles",
    label_key="label",
    epochs=50,
    batch_size=64,
    learning_rate=5e-5,
    early_stopping_patience=5,
)


multi_results = runner_std.run_multi(
    run_seeds=[42, 43, 44, 45, 46],
    fold_seeds=[0, 1,2, 3, 4],
    split_type="ecfp_split",
)

print(multi_results.to_dataframe().head())
print(multi_results.get_summary_stats().get("test_roc-auc"))
```

# 2. non-canonical peptide regression


```python
from pepbenchmark.model.smiles_model import build_smiles_model
from pepbenchmark.dataset_manager.single_dataset import SinglePeptideDatasetManager


dataset = SinglePeptideDatasetManager(
    "nc-cpp_pampa",
    official_feature_names=["smiles", "label"],
    dataset_dir="../../PepBenchData/PepBenchData-50",
)
runner = build_smiles_model(
    dataset=dataset,
    model_name="seyonec/ChemBERTa-zinc-base-v1",
    task_type="regression",
    smiles_key="smiles",
    label_key="label",
    epochs=50,
    batch_size=64,
    learning_rate=5e-5,
    early_stopping_patience=5,
)

multi_results = runner.run_multi(
    run_seeds=[42, 43, 44, 45, 46],
    fold_seeds=[0, 1, 2, 3, 4],
    split_type="ecfp_split",   
)
print(multi_results.to_dataframe())
summary = multi_results.get_summary_stats()
print(summary["test_mae"])
```
